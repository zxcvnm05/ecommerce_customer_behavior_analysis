## **VII. Check invalid values and outliers**
Check invalid values and detect outliers using the IQR method.

```python
print("Invalid value checks:")

invalid_age = df[(df["customer_age"] < 18) | (df["customer_age"] > 100)]
invalid_order_value = df[df["order_value_usd"] <= 0]
invalid_delivery_time = df[
    (df["delivery_time_days"] <= 0) | (df["delivery_time_days"] > 30)
]
invalid_rating = df[
    (df["customer_rating"] < 1) | (df["customer_rating"] > 5)
]

print("Invalid customer_age count:", len(invalid_age))
print("Invalid order_value_usd count:", len(invalid_order_value))
print("Invalid delivery_time_days count:", len(invalid_delivery_time))
print("Invalid customer_rating count:", len(invalid_rating))


print("\nPotential outlier detection using IQR method:")

numeric_cols_for_outlier = [
    "customer_age",
    "order_value_usd",
    "delivery_time_days",
    "customer_rating"
]

outlier_summary = []

for col in numeric_cols_for_outlier:
    series = df[col].dropna()

    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    number_of_outliers = df[
        (df[col] < lower_bound) | (df[col] > upper_bound)
    ].shape[0]

    outlier_summary.append({
        "variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "number_of_outliers": number_of_outliers
    })

outlier_summary_df = pd.DataFrame(outlier_summary)
display(outlier_summary_df)
```

# **B. DATA PREPROCESSING**

## **I. Standardize column names**

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Column names after standardization:")
print(df.columns.tolist())
```

**Column names are standardized to lowercase with underscores**

## **II. Remove duplicate records**
Remove duplicate records so each transaction appears only once.

```python
rows_before_duplicates = len(df)

df = df.drop_duplicates()

rows_after_duplicates = len(df)

print("Rows before removing duplicates:", rows_before_duplicates)
print("Rows after removing duplicates:", rows_after_duplicates)
print("Number of duplicate rows removed:", rows_before_duplicates - rows_after_duplicates)

print("Duplicated order_id after removing duplicate rows:", df["order_id"].duplicated().sum())
```

## **III. Standardize categorical variables**
Standardize categorical variables to remove inconsistent capitalization.

```python
def clean_text(value):
    if pd.isna(value):
        return np.nan
    return str(value).strip().title()

categorical_columns = [
    "customer_gender",
    "product_category",
    "payment_method",
    "returned"
]

for col in categorical_columns:
    df[col] = df[col].apply(clean_text)

df["payment_method"] = df["payment_method"].replace({
    "Upi": "UPI",
    "Cash On Delivery": "Cash on Delivery"
})

print("Unique values after standardizing categorical variables:")

for col in categorical_columns:
    print("\n" + col)
    print(df[col].value_counts(dropna=False))
```

**Values such as electronics, ELECTRONICS are standardized to Electronics; yes, NO are standardized to Yes, No.**

## **IV. Convert order_date to datetime**
Convert order_date from text/object to datetime for time-based analysis.

```python
def parse_mixed_order_date(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    if "-" in value:
        return pd.to_datetime(value, errors="coerce", dayfirst=True)

    if "/" in value:
        return pd.to_datetime(value, errors="coerce", dayfirst=False)

    return pd.to_datetime(value, errors="coerce")

df["order_date"] = df["order_date"].apply(parse_mixed_order_date)

print("order_date after conversion:")
print("Data type:", df["order_date"].dtype)
print("Minimum date:", df["order_date"].min())
print("Maximum date:", df["order_date"].max())
print("Missing dates after conversion:", df["order_date"].isnull().sum())

print("\nSample converted dates:")
display(df[["order_id", "order_date"]].head(20))
```

## **V. Handle invalid numerical values**
Convert invalid numerical values into missing values before imputation.

```python
invalid_rating_mask = (df["customer_rating"] < 1) | (df["customer_rating"] > 5)
print("Invalid customer_rating values converted to NaN:", invalid_rating_mask.sum())
df.loc[invalid_rating_mask, "customer_rating"] = np.nan

invalid_delivery_mask = (df["delivery_time_days"] <= 0) | (df["delivery_time_days"] > 30)
print("Invalid delivery_time_days values converted to NaN:", invalid_delivery_mask.sum())
df.loc[invalid_delivery_mask, "delivery_time_days"] = np.nan

invalid_value_mask = df["order_value_usd"] <= 0
print("Invalid order_value_usd values converted to NaN:", invalid_value_mask.sum())
df.loc[invalid_value_mask, "order_value_usd"] = np.nan
```

## **VI. Cap outliers in order_value_usd**
Reduce the influence of extreme order_value_usd values using IQR capping

```python
Q1 = df["order_value_usd"].quantile(0.25)
Q3 = df["order_value_usd"].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

outlier_order_value_count = (df["order_value_usd"] > upper_bound).sum()

print("Order value upper bound using IQR:", round(upper_bound, 2))
print("Order value outliers capped:", outlier_order_value_count)

df["order_value_usd"] = np.where(
    df["order_value_usd"] > upper_bound,
    upper_bound,
    df["order_value_usd"]
)
```

## **VII. Handle missing values**
Impute missing values using suitable methods: mode for categorical variables and median for numerical variables.

```python
print("Missing values before imputation:")
display(df.isnull().sum())

payment_mode = df["payment_method"].mode(dropna=True)[0]
df["payment_method"] = df["payment_method"].fillna(payment_mode)
print("Missing payment_method filled with mode:", payment_mode)

delivery_median = df["delivery_time_days"].median()
df["delivery_time_days"] = df["delivery_time_days"].fillna(delivery_median)
print("Missing delivery_time_days filled with median:", delivery_median)

rating_median = df["customer_rating"].median()
df["customer_rating"] = df["customer_rating"].fillna(rating_median)
print("Missing customer_rating filled with median:", rating_median)

missing_dates = df["order_date"].isnull().sum()
print("Missing order_date after date conversion:", missing_dates)

if missing_dates > 0:
    df = df.dropna(subset=["order_date"])

print("\nMissing values after imputation:")
display(df.isnull().sum())
```

## **VIII. Convert data types after cleaning**
Convert variables into appropriate data types after cleaning.

```python
df["delivery_time_days"] = df["delivery_time_days"].round().astype(int)
df["customer_age"] = df["customer_age"].astype(int)
df["order_id"] = df["order_id"].astype(int)

print("Data types after cleaning:")
display(df.dtypes)
```

# **C. FEATURE ENGINEERING**



## **I. Create returned_flag**
Convert returned from Yes/No to 1/0 for return-rate analysis

```python
df["returned_flag"] = df["returned"].map({
    "Yes": 1,
    "No": 0
})

print("Returned flag distribution:")
display(df["returned_flag"].value_counts(dropna=False))

print("\nOverall return rate:")
print(round(df["returned_flag"].mean() * 100, 2), "%")
```

## **II. Create age_group**
Segment customers into age groups for customer behavior analysis.

```python
age_bins = [17, 25, 35, 45, 55, 70]
age_labels = ["18-25", "26-35", "36-45", "46-55", "56-69"]

df["age_group"] = pd.cut(
    df["customer_age"],
    bins=age_bins,
    labels=age_labels
)

print("Age group distribution:")
display(df["age_group"].value_counts().sort_index())
```

## **III. Create order_value_segment**
Classify orders into low, medium, and high value segments.

```python
df["order_value_segment"] = pd.qcut(
    df["order_value_usd"],
    q=3,
    labels=["Low value", "Medium value", "High value"]
)

print("Order value segment distribution:")
display(df["order_value_segment"].value_counts())
```

**This variable supports return-rate comparison by spending level.**

## **IV. Create delivery_speed**
Classify delivery time into fast, standard, and slow delivery.

```python
def categorize_delivery_speed(days):
    if days <= 3:
        return "Fast delivery"
    elif days <= 7:
        return "Standard delivery"
    else:
        return "Slow delivery"

df["delivery_speed"] = df["delivery_time_days"].apply(categorize_delivery_speed)

print("Delivery speed distribution:")
display(df["delivery_speed"].value_counts())
```

**The delivery_speed variable supports analysis of delivery performance, rating, and return rate**

## **V. Create time-based variables**
Create time-based variables for yearly, monthly, and quarterly trend analysis

```python
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["order_quarter"] = df["order_date"].dt.quarter
df["order_month_year"] = df["order_date"].dt.to_period("M").astype(str)

print("Time-based variables created:")
display(df[[
    "order_date",
    "order_year",
    "order_month",
    "order_quarter",
    "order_month_year"
]].head())
```

**RESULT**

---
| Variable           | Purpose                              |
| ------------------ | ------------------------------------ |
| `order_year`       | Yearly trend analysis                |
| `order_month`      | Monthly trend analysis               |
| `order_quarter`    | Quarterly trend analysis             |
| `order_month_year` | Month-year time-series visualization |



## **VI. Create is_high_value_order**
Identify high-value orders based on the 75th percentile of order_value_usd

```python
high_value_threshold = df["order_value_usd"].quantile(0.75)

df["is_high_value_order"] = np.where(
    df["order_value_usd"] >= high_value_threshold,
    1,
    0
)

print("High-value order threshold:", round(high_value_threshold, 2))

print("\nHigh-value order distribution:")
display(df["is_high_value_order"].value_counts())
```
